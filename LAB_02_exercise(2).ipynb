```python
# Task (ii): Make a multiplication table using a loop (Table of 5)
num = 5
for i in range(1,11):
    print (f"{num} x {i} = {num * i}")
```

```python
# Task (iii): Output the numbers 1 to 10 backwards using a loop
for i in range (10,0,-1):
    print (i)
```

```python
# Task (v): Create a loop that sums the numbers from 100 to 200
total = 0
for i in range (100, 201):
    total += i
print ("The sum is:",total)
```
