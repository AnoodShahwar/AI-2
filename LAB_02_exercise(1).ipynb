```python
# (i): Caluclate the volume of cube
```

```python
def calculate_volume ():
    height = float(input("Enter height of the cube in cm:\n"))
    width = float(input("Enter width of the cube in cm:\n"))
    depth = float(input("Enter depth of the cube in cm:\n"))

    volume = height * width * depth
    print("\nVolume is:\n", volume, "cm³")

    if (1 <= volume <= 10) :
        label= "Extra Small"
    elif (11 <= volume <= 25) :
        label= "Small"
    elif (26 <= volume <= 75) :
        label= "Medium"
    elif (76 <= volume <= 100) :
        label= "Large"
    elif (101 <= volume <= 250) :
        label= "Extra Large"
    elif (volume > 251) :
        label= "Extra-Extra Large"
    else:
        label= "Invalid"
        
    print ("\nLabel:\n", label)

calculate_volume()
```
